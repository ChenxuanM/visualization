```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
from PIL import Image
import requests

# Set the random seed to ensure reproducibility
np.random.seed(613)

# Generate data
x = np.arange(50)
y1 = np.random.randint(0, 100, 50)
y2 = np.random.randint(0, 100, 50)

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label="Person 1")
ax.plot(x, y2, label="Person 2")
# Place legend in the lower right corner
ax.legend(loc='lower right')
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label="Person 1")
ax.plot(x, y2, label="Person 2")
ax.legend(loc='lower right',
          frameon=True,  # Show frame/border
          fontsize=12,   # Font size
          ncol=2,        # Two columns
          shadow=True)   # Add shadow
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x, y1, label="Person 1")
ax.plot(x, y2, label="Person 2")
# (1, 1) places the legend at the top-right corner outside the plot
ax.legend(loc='upper left', bbox_to_anchor=(1, 1))
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')
ax.text(10, 95, "This value is important!",
        ha='center',
        color='red',
        size=20)
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')

ax.annotate('This is important!',
            xy=(10, 95),
            xytext=(20, 94),
            arrowprops=dict(arrowstyle="wedge", color="pink"))
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")

# Hide all ticks on the Y-axis
ax.yaxis.set_major_locator(plt.NullLocator())

# Hide only the text labels on the X-axis (keep the tick lines)
ax.xaxis.set_major_formatter(plt.NullFormatter())
plt.show()
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")

# Hide all ticks on the Y-axis
ax.yaxis.set_major_locator(plt.NullLocator())

# Hide only the text labels on the X-axis (keep the tick lines)
ax.xaxis.set_major_formatter(plt.NullFormatter())

# Define font dictionary
font1 = {'family': 'serif', 'color': 'indigo'}

fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x, y1, label="Person 1")
ax.scatter(x, y2, label="Person 2")
ax.legend(loc='lower right')

# Apply font style to X-axis label
plt.xlabel('newnewnewx!', fontsize=18, fontdict=font1)
plt.show()


```
