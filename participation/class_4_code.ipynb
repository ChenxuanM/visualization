```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
from PIL import Image
import requests
```

```python
# 1. Set Random Seed
np.random.seed(613)

# 2. Generate Data Set 1 (Numerical)
x1 = np.arange(50)
y1 = np.random.randint(0, 75, 50)

# 3. Generate Data Set 2 (Categorical - One Piece Characters)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])

print("Setup Complete! Data generated.")
```

```python
###Basic Subplots
# Create a figure with 1 row and 2 columns
fig, (ax1, ax2) = plt.subplots(ncols=2, nrows=1, figsize=(7, 3))

# Plot Scatter on the first axis (left)
ax1.scatter(x1, y1)
ax1.set_title("Scatter Plot")

# Plot Bar on the second axis (right)
ax2.bar(x2, y2)
ax2.set_title("Bar Chart")

# Define the layout: 'ax1' and 'ax2' on the left, 'ax3' on the right (spanning top and bottom)
# layout="constrained" automatically prevents overlapping labels
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize=(7, 4),
                                    layout="constrained")
# Show the plot
plt.show()
```

```python
# Define the layout: 'ax1' and 'ax2' on the left, 'ax3' on the right (spanning top and bottom)
# layout="constrained" automatically prevents overlapping labels
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize=(7, 4),
                                    layout="constrained")

# Plot data onto specific axes using the keys defined above
someaxes["ax1"].scatter(x1, y1)
someaxes["ax2"].bar(x2, y2)
someaxes["ax3"].plot(x1, y1)

# Add large labels to demonstrate automatic layout adjustment
someaxes["ax1"].set_xlabel('Frist title', fontsize=18)
someaxes["ax2"].set_xlabel('Second title', fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18)
```

```python
# --- 1. Re-define data for this section (as per slides) ---
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y_bar = np.array([110, 180, 240, 99, 220])
y_line = np.array([170, 100, 90, 120, 50])

# Calculate Standard Deviation for error bars
y_sd = np.std(y_line)

# --- 2. Plotting ---
fig, ax = plt.subplots(figsize=(7, 3))

# Plot Bar chart
ax.bar(x, y_bar, color="indigo", label="Bar Data")

# Plot Line chart on top
ax.plot(x, y_line, color="red", label="Line Data")

# --- 3. Add Error Bars ---
# fmt='none' means we don't plot the points again, just the error bars
# errorevery=2 means show error bars on every 2nd point (optional)
ax.errorbar(x, y_line,
            yerr=y_sd,
            fmt="none",
            ecolor="indigo",
            elinewidth=4,
            capsize=6,
            capthick=4,
            errorevery=1) # Set to 1 to show all, or 2 to skip

ax.legend()
plt.show()
```

```python
from PIL import Image
import requests
from io import BytesIO

url = 'https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png'

headers = {'User-Agent': 'Mozilla/5.0'}
response = requests.get(url, headers=headers)

image_file = BytesIO(response.content)
image = Image.open(image_file)
image
```

```python
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y_bar = np.array([110, 180, 240, 99, 220])
y_line = np.array([170, 100, 90, 120, 50])

fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x, y_line, color="red")
ax.set_title("Power Levels with Image")

# 3. Add a new axis for the image
# [x, y, width, height] in figure coordinates (0 to 1)
ax_image = fig.add_axes([0.1, 0.11, 0.15, 0.35])

# 4. Show the image and turn off axis lines
ax_image.imshow(image)
ax_image.axis('off')

plt.show()
```
