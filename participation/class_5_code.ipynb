```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# 1. Load the 'tips' dataset (built-in sample data)
tips = sns.load_dataset("tips")
print(tips.head()) # Preview the data

# Basic Line Plot
# Plotting 'tip' amount against 'total_bill'
sns.lineplot(data=tips, x='total_bill', y='tip')
plt.show()

# Apply a Style
# Options: 'darkgrid', 'whitegrid', 'dark', 'white', 'ticks'
sns.set_style('whitegrid')

# Add Titles and Labels
tipgraph = sns.lineplot(data=tips, x='total_bill', y='tip')
tipgraph.set(title='Tips vs. Total Bill',
             xlabel='Total Bill ($)',
             ylabel='Tip Amount ($)')
plt.show()
```

```python
# 1. Complex Scatter Plot (Multivariate)
# hue='day': Color points by day
# style='time': Shape points by time (Lunch/Dinner)
plt.figure(figsize=(10, 6))
sns.scatterplot(data=tips,
                x='total_bill',
                y='tip',
                style='time',
                hue='day',
                palette=['purple', 'hotpink', 'deepskyblue', 'green'])
plt.title('Tips vs. Total Bill by Day and Time')
plt.show()
```

```python

# 2. Pairplot (Quickly compare all variables)
sns.pairplot(data=tips, hue='day')
plt.show()
```

```python

# 3. Relplot (Facet Grid - Separate plots by column)
# col='day': Create a separate plot for each day
sns.relplot(data=tips,
            x="total_bill",
            y="tip",
            hue="sex",
            col="day",
            kind="scatter",
            col_wrap=2)
plt.show()
```

```python
import plotly.graph_objects as go
import numpy as np

# 1. Generate Data
x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])

# 2. Create an Interactive Bar Chart
graph = go.Figure()
graph.add_trace(go.Bar(x=x1, y=y1))

# Update layout (Titles)
graph.update_layout(title="Pirate Scores",
                    xaxis_title="Pirates",
                    yaxis_title="Score")
graph.show()

```

```python
graph2 = go.Figure()
graph2.add_trace(go.Scatter(x=x1, y=y1,
                            mode='markers', # 'lines' or 'markers'
                            marker=dict(size=15,
                                        color='blue',
                                        opacity=1,
                                        line=dict(width=5, color='purple'))
                           ))
graph2.update_layout(title='Interactive Pirate Plot', width=500, height=500)
graph2.show()
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt
import pandas as pd


df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv", on_bad_lines='skip')


text = " ".join(each for each in df.quote)


# background_color='white': Set background to white
# colormap='inferno': Color scheme
wordcloud = WordCloud(background_color="white", colormap='inferno').generate(text)


fig, ax = plt.subplots(figsize=(10, 5))
ax.imshow(wordcloud, interpolation='bilinear')
ax.axis("off") # Hide axis
plt.show()
```

```python
from matplotlib_venn import venn2, venn2_unweighted
import matplotlib.pyplot as plt


A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])


plt.figure(figsize=(6, 6))
diagram = venn2_unweighted([A, B],
                           set_labels=('Fruits', 'Pies'),
                           set_colors=("blue", "red"),
                           alpha=0.5)


# '10': Items only in A
# '11': Items in both A and B (Intersection)
# '01': Items only in B
# "\n".join(...) puts each item on a new line
diagram.get_label_by_id("10").set_text("\n".join(A - B))
diagram.get_label_by_id("11").set_text("\n".join(A & B))
diagram.get_label_by_id("01").set_text("\n".join(B - A))

plt.title("Venn Diagram of Fruits and Pies")
plt.show()
```
